```python


import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Student_Performance_Prediction (1).csv")

print("Column Names:")
print(df.columns.tolist())

samples, features = df.shape

print("\nNumber of Samples:", samples)
print("Number of Features:", features)

print("\nFirst 5 Rows:")
print(df.head())

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Study_Hours", y="Final_Score")

plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# Split the data into reproducible training and testing sets.
train_df = df.sample(frac=0.8, random_state=42)
test_df = df.drop(train_df.index)

X_train = train_df.drop(columns=["Final_Score"])
y_train = train_df["Final_Score"]
X_test = test_df.drop(columns=["Final_Score"])
y_test = test_df["Final_Score"]

print("Training features:", X_train.shape)
print("Training target:", y_train.shape)
print("Testing features:", X_test.shape)
print("Testing target:", y_test.shape)
```
