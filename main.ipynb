```python
import numpy as np

data = np.array([[10, -20, 30],
                  [40, -50, 60],
                  [70, 80, -90]])

print("Data:\n", data)
print("\nMean of each row:      ", data.mean(axis=1))
print("Mean of each column:   ", data.mean(axis=0))
print("Std dev of each row:   ", data.std(axis=1))
print("Max of each column:    ", data.max(axis=0))
print("Min of each row:       ", data.min(axis=1))
```
